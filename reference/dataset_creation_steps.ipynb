# Changing the Folder and File formats of 30 New Datasets

```python
resample_path = Path(r"C:\Users\Gowtham\Downloads\30NewDatasets")
DRY_RUN = True  # set to False once the printed renames look right

# Keep the _TRAIN / _TEST split suffix intact (loaders expect it)
SPLIT_SUFFIX = re.compile(r"_(TRAIN|TEST)(?=\.|$)")


def camelize(name: str) -> str:
    """AconityMINIPrinterLarge_eq -> AconityMINIPrinterLargeEq
    AconityMINIPrinterLarge_eq_TRAIN.ts -> AconityMINIPrinterLargeEq_TRAIN.ts"""
    suffix = ""
    m = SPLIT_SUFFIX.search(name)
    if m:
        suffix = name[m.start() :]
        name = name[: m.start()]
    parts = name.split("_")
    if len(parts) > 1 and parts[1] == "disc":
        parts[1] = "discrete"
    new = parts[0] + "".join(p[:1].upper() + p[1:] for p in parts[1:])
    return new + suffix


def rename(path: Path):
    new_name = camelize(path.name)
    if new_name == path.name:
        return
    target = path.with_name(new_name)
    if target.exists():
        print(f"SKIP (target exists): {path.name} -> {new_name}")
        return
    print(f"{path.name}  ->  {new_name}")
    if not DRY_RUN:
        path.rename(target)


# Bottom-up so files are renamed before their parent folders
for root, dirs, files in os.walk(resample_path, topdown=False):
    root = Path(root)
    for f in files:
        rename(root / f)
    for d in dirs:
        rename(root / d)
```

### Script it out

```python
base_dir = Path(r"C:\Users\Gowtham\Downloads\30NewDatasets")

def generate_resamples(base_dir: str = base_dir):
    try:
        for folder in folder_names:
            base_train_file_path = os.path.join(base_dir, folder, f"{folder}_TRAIN.ts")
            base_test_file_path = os.path.join(base_dir, folder, f"{folder}_TEST.ts")
            
            base_X_train, base_y_train = load_from_tsfile(base_train_file_path)
            base_X_test, base_y_test = load_from_tsfile(base_test_file_path)
            
            print(f"\nWORKING ON DATASET {folder}\n---")
            for resample_id in range(0, 30):
                if resample_id == 0:
                    X_train_resample, y_train_resample, X_test_resample, y_test_resample = base_X_train, base_y_train, base_X_test, base_y_test
                else:
                    X_train_resample, y_train_resample, X_test_resample, y_test_resample = stratified_resample_data(
                        base_X_train, base_y_train, base_X_test, base_y_test, random_state=resample_id
                    )
                
                base_folder_path = os.path.join(base_dir, folder)
                write_to_tsfile(X=X_train_resample, y=y_train_resample, path=base_folder_path, problem_name=f"{folder}{resample_id}_TRAIN.ts")
                write_to_tsfile(X=X_test_resample, y=y_test_resample, path=base_folder_path, problem_name=f"{folder}{resample_id}_TEST.ts")
                print(f"Generated {resample_id + 1}/30 resamples.")
    except Exception as e:
        print(f"Exception Occurred: {e}")
```

# Sanity Check

To check if my resampled dataset goes along with the resample indices given by them

```python
base_dir = Path(r"C:\Users\Gowtham\Downloads\30NewDatasets")
resample_indices_path = Path(r"C:\Users\Gowtham\Downloads\PythonResampleIndices\PythonResampleIndices")

def check_resample_equality(dataset: str, resample_id: int):
    base_train_file_path = os.path.join(base_dir, dataset, f"{dataset}_TRAIN.ts")
    base_test_file_path = os.path.join(base_dir, dataset, f"{dataset}_TEST.ts")
    resample_indices_train_file_path = os.path.join(resample_indices_path, dataset, f"resample{resample_id}Indices_TRAIN.txt")
    resample_indices_test_file_path = os.path.join(resample_indices_path, dataset, f"resample{resample_id}Indices_TEST.txt")
    resampled_train_file_path = os.path.join(base_dir, dataset, f"{dataset}{resample_id}_TRAIN.ts")
    resampled_test_file_path = os.path.join(base_dir, dataset, f"{dataset}{resample_id}_TEST.ts")

    X_train_base, y_train_base = load_from_tsfile(base_train_file_path)
    X_test_base, y_test_base = load_from_tsfile(base_test_file_path)

    X_base, y_base = np.concatenate([X_train_base, X_test_base], axis=0), np.concatenate([y_train_base, y_test_base], axis=0)

    X_train_res, y_train_res = load_from_tsfile(resampled_train_file_path)
    X_test_res, y_test_res = load_from_tsfile(resampled_test_file_path)

    train_f = open(resample_indices_train_file_path)
    train_indices = [int(line.removesuffix("\n")) for line in train_f.readlines()]

    test_f = open(resample_indices_test_file_path)
    test_indices = [int(line.removesuffix("\n")) for line in test_f.readlines()]
    
    X_train_res_base, y_train_res_base = X_base[train_indices, :, :], y_base[train_indices]
    X_test_res_base, y_test_res_base = X_base[test_indices, :, :], y_base[test_indices]
    
    assert np.array_equal(X_train_res_base, X_train_res)
    assert np.array_equal(X_test_res_base, X_test_res)
    assert np.array_equal(y_train_res_base, y_train_res)
    assert np.array_equal(y_test_res_base, y_test_res)
```

```python
for dataset in folder_names:
    for resample_id in range(0, 30):
        check_resample_equality(dataset, resample_id)
    print(f"The resamples work perfectly for {dataset}.")
```
